```python
# Boucle pour activer les liens url et lancer le téléchargement du fichier détaillé
# Objectif: enregistrer les fichiers .csv dans un même sous-dossier

# 1. enregistrer les url dans une liste
liste_url= df["URL"].tolist()
print(liste_url)

# 2. enregistrer les fichiers .zip dans le sous-dossier data_freq_details
import requests
import os

dossier= r"C:\Users\justi\OneDrive\Je-deviens-Data-Analyst\JEDHA\00_Certif\bloc_6\02_cleaning_python\data_freq_details"

for url in liste_url:
    nom_fichier= os.path.basename(url)
    chemin_complet= os.path.join(dossier, nom_fichier)
    print(f"Téléchargement de {nom_fichier}...")

    fichier= requests.get(url)
    if fichier.status_code == 200:
        with open(chemin_complet, "wb") as f:       # ouvre le fichier en sachant qu'il va devoir écrire ("w") dedans et que c'est du binaire ("b")
            f.write(fichier.content)                # écrit dans le fichier qu'il a ouvert, le contenu du fichier téléchargé
        print(f"...{nom_fichier} enregistré dans {dossier}.")
    else:
        print(f"Erreur {fichier.status_code} pour {url}.")
```

```python
# Boucle pour dézipper les .zip

import zipfile
import glob

# Définir les chemins:
dossier_avec_zip= r"C:\Users\justi\OneDrive\Je-deviens-Data-Analyst\JEDHA\2_Fullstack\11_final_project\data_freq_details"
dossier_destination= r"C:\Users\justi\OneDrive\Je-deviens-Data-Analyst\JEDHA\2_Fullstack\11_final_project\data_freq_details\csv"

# Trouver tous les fichiers .zip dans le dossier:
fichiers_zip= glob.glob(f"{dossier_avec_zip}/*.zip")    # j'ajoute / pour que le chemin fonctionne
                                                        # la bonne pratique serait d'utiliser os.path.join 

if not fichiers_zip:
    print(f"Aucun fichier .zip trouvé dans le dossier : {dossier_avec_zip}")
else:
    print(f"{len(fichiers_zip)} fichiers .zip trouvés. Début de l'extraction...")   # Permet de savoir si Python trouve les fichiers

# Boucle sur chaque fichier pour l'extraire:
    for fichier in fichiers_zip:
        with zipfile.ZipFile(fichier, "r") as zip_ref:      #'r' to read an existing file
            zip_ref.extractall(dossier_destination)
            print(f"Extrait {fichier} ok.")
```

```python
# la boucle s'arrête du fait d'une erreur, je recommence avec un Try/Except

import zipfile
import glob

dossier_avec_zip= r"C:\Users\justi\OneDrive\Je-deviens-Data-Analyst\JEDHA\2_Fullstack\11_final_project\data_freq_details"
dossier_destination= r"C:\Users\justi\OneDrive\Je-deviens-Data-Analyst\JEDHA\2_Fullstack\11_final_project\data_freq_details\csv"

fichiers_zip= glob.glob(f"{dossier_avec_zip}/*.zip")    

if not fichiers_zip:
    print(f"Aucun fichier .zip trouvé dans le dossier : {dossier_avec_zip}")
else:
    print(f"{len(fichiers_zip)} fichiers .zip trouvés. Début de l'extraction...")   

# Boucle sur chaque fichier pour l'extraire:
    for fichier in fichiers_zip:
        try:
            with zipfile.ZipFile(fichier, "r") as zip_ref:      #'r' to read an existing file
                zip_ref.extractall(dossier_destination)
                print(f"Extrait {fichier} ok.")
        
        except zipfile.BadZipFile:
            print(f"ERREUR: le fichier {fichier} est corrompu ou n'est pas un fichier zip valide.")
        except Exception:
            print(f"ERREUR lors du traitement de {fichier}: {Exception}")

print("Processus de dézippage terminé.")
```

```python
# Boucle pour ouvrir les fichiers .csv, les convertir en df avec pandas et ajouter les lignes dans un seul et même df avec concat
# Ajout du mask pour garder uniquement les communes du MVP et réduire le volume de lignes
# Je vais essayer d'isoler la chaine moisannée de la date avec un split

import os

chemin = r"C:\Users\justi\OneDrive\Je-deviens-Data-Analyst\JEDHA\00_Certif\bloc_6\02_cleaning_python\data_freq_details\csv"

fichiers = [f for f in os.listdir(chemin) 
            if os.path.isfile(os.path.join(chemin, f))]

fichiers = sorted(fichiers, reverse=True)

LastYear = fichiers[-12:]

print(LastYear[:3])

frequentations= []

for fichier in LastYear:
    df= pd.read_csv(fichier, sep= ';')
    mask= (df['NomCommune'] != 'Rennes')
    df= df.loc([mask],['DateFreq', 'TrancheHoraire15mn', 'Timeo', 'NomArret', 'NomCommune',
       'identifiantLigne', 'NomCourtLigne', 'Sens', 'Frequentation'])
    df.rename(columns={
    'DateFreq' : 'date', 
    'TrancheHoraire15mn' : 'tranche_horaire', 
    'Timeo' : 'id_arret', 
    'NomCommune' : 'nom_commune',
    'identifiantLigne' : 'id_ligne', 
    'Sens' : 'sens', 
    'Frequentation': 'freq'
    }, inplace= True)
    df['date'] = pd.to_datetime(df['date'], dayfirst= True)
    df['tranche_horaire'] = pd.to_datetime(df['tranche_horaire'], format="%H:%M:%S").dt.time
    df= df.astype({
    'id_arret' : 'int64', 
    'nom_commune' : 'string', 
    'id_ligne' : 'int64',
    'sens' : 'string'
    })
    frequentations.append(df)

print(frequentations[:3])

```
