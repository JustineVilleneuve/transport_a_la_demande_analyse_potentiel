```python
# Boucle pour ouvrir les fichiers .csv, les convertir en df avec pandas et ajouter les lignes dans un seul et même df avec concat
# Ajout du mask pour garder les communes du MVP et réduire le volume de lignes
# Je vais essayer d'isoler la chaine moisannée de la date avec un split
# On décide de se concentrer sur les 12 derniers mois disponibles dans le cadre du MVP, afin de réduire le nombre de lignes

import os

chemin = r"C:\Users\justi\OneDrive\Je-deviens-Data-Analyst\JEDHA\2_Fullstack\11_final_project\.old\data_freq_details\csv"

fichiers = [f for f in os.listdir(chemin) 
            if os.path.isfile(os.path.join(chemin, f))]

fichiers = sorted(fichiers, reverse=False)

LastYear = fichiers[-12:]

print(LastYear)
```

```python
colonnes= ['DateFreq', 'TrancheHoraire15mn', 'Timeo', 'NomCommune',
       'identifiantLigne', 'Sens', 'Frequentation']

colonnes_def= {
    'DateFreq' : 'date', 
    'TrancheHoraire15mn' : 'tranche_horaire', 
    'Timeo' : 'id_arret', 
    'NomCommune' : 'nom_commune',
    'identifiantLigne' : 'id_ligne', 
    'Sens' : 'sens', 
    'Frequentation': 'freq'
    }

dtypes= {
    'id_arret' : 'Int64', 
    'nom_commune' : 'string', 
    'id_ligne' : 'Int64',
    'sens' : 'string'
}

frequentations= []

for fichier in LastYear:
    
    df= pd.read_csv(
        os.path.join(chemin, fichier),
        sep= ';',
        usecols= colonnes,
        parse_dates= ['DateFreq'],
        dayfirst= True
        )
    
    df= df.rename(columns= colonnes_def)

    df= df.loc[df['nom_commune'] != 'Rennes']

    df['tranche_horaire'] = pd.to_datetime(
        df['tranche_horaire'], format="%H:%M:%S"
        ).dt.time

    df= df.astype(dtypes)
    
    frequentations.append(df)

print(frequentations[:3])
```

```python
df_concat= pd.concat(frequentations, ignore_index= True)

print(df_concat.shape)
display(df_concat.describe(include="all"))
display(df_concat.isna().sum())

df_concat
```
