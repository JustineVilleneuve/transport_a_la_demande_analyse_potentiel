```python
#Création de la colonne 'MVP' pour cibler les communes de la périphérie de Rennes et filtrer le dataset dessus
MVP = [35032,35058,35196,35065,35275,35144,35266,35315,35278,35245,35208,35353,35120,35210,35352,35059,35250,35079,35363,35051,35055,35076,35189,35066,35204,35022,35351,35088,35080,35131,35001,35206,35081,35139,35334,35216,35024,35240,35039,35047,35281,35180]
bus_arrets['MVP'] = bus_arrets['Commune (code INSEE)'].apply(lambda x: 'oui' if x in MVP else 'non')
bus_arrets['MVP'].describe()
```

```python
#Création du mask pour filtrer le dataset sur les communes sélectionnées dans la périphérie de Rennes
mask = (bus_arrets['MVP'] == 'oui')
bus_arrets = bus_arrets[mask]

bus_arrets.shape
```

```python
# Split de la colonne coordonnées en latitude et longitude
bus_arrets[['latitude', 'longitude']] = bus_arrets['Coordonnées'].str.split(',', expand=True)
bus_arrets.head()
```

```python
# Selection des colonnes utiles
keep_cols = ['Identifiant','Nom', 'Commune (code INSEE)','latitude', 'longitude']
df = bus_arrets.loc[:,keep_cols]
df.head()
```

```python
# Renommage des colonnes gardées
cols_new = ["id_arret", "nom_arret", "code_geo","latitude", "longitude"]
df.columns = cols_new
df.head()
```

```python
# Vérification des types des colonnes
df.dtypes
```

```python
df.columns
```

```python
# Changement de types pour certaines colonnes
df = df.astype({
    'nom_arret': 'string',
    'latitude' : 'float64',
    'longitude' : 'float64'
})
df.dtypes
```
