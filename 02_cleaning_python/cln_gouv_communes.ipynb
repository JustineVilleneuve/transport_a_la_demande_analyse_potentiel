```python
# Filtrage pour garder le département de l'Ille-et-Vilaine et identification des colonnes à conserver
mask= (df['code_insee'].str[:2] == "35")
cols =['code_insee', 'nom_standard', 'codes_postaux', 'canton_code', 'canton_nom', 'superficie_hectare', 'superficie_km2', 'latitude_centre', 'longitude_centre']

df= df.loc[mask, cols]
df.head()
```

```python
# Création de la colonne 'MVP' pour cibler les communes cibles et filtrer le dataset dessus
MVP = [35032, 35058, 35196, 35065, 35275, 35144, 35266, 35315, 35278, 35245, 35208, 35353, 35120, 35210, 35352, 35059, 35250, 35079, 35363, 35051, 35055, 35076, 35189, 35066, 35204, 35022, 35351, 35088, 35080, 35131, 35001, 35206, 35081, 35139, 35334, 35216, 35024, 35240, 35039, 35047, 35281, 35180]

df['MVP'] = df['code_insee'].astype(int).apply(lambda x: 'oui' if x in MVP else 'non')

df['MVP'].describe()
```

```python
# Filtrage du dataset pour conserver que les communes de notre MVP:

mask= (df['MVP'] == 'oui')
df= df.loc[mask, cols]
df
```

```python
# Reste à renommer les colonnes et les passer au format souhaité
df.rename(columns={'code_insee' : 'code_geo', 'nom_standard' : 'nom_commune'}, inplace= True)

df['code_geo']= df['code_geo'].astype(int)

df.dtypes
```

```python
# Exporter le dataset nettoyé en csv
df.to_csv("stg_gouv_communes.csv", sep=";", index=False, encoding='UTF-8')
```
