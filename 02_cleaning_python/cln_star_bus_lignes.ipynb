```python
#Librairie(s) utilisée(s)
import pandas as pd
```

```python
#Création du dataframe à partir du fichier csv
bus_lignes = pd.read_csv(r"C:\Users\justi\OneDrive\Je-deviens-Data-Analyst\JEDHA\00_Certif\bloc_6\01_data\01_data_bronze\tco-bus-topologie-lignes-td.csv"
                         , delimiter=";")
bus_lignes.head()
```

```python
# Nombre de lignes avec valeur
bus_lignes.count()
```

```python
# Valeurs manquantes
bus_lignes.isnull().sum()
```

```python
bus_lignes.columns
```

```python
# Selection des colonnes utiles
keep_cols = ['Identifiant','Est la version active','Nom court', 'Famille commerciale' ]
df = bus_lignes.loc[:,keep_cols]
df.head()
```

```python
# Vérification si tout est la version active
df['Est la version active'].unique()
```

```python
# Suppression de cette colonne
df = df.drop('Est la version active', axis=1)
df.head()
```

```python
# Renommage des colonnes gardées
df.columns = ['id_ligne', 'nom_ligne', 'famille_ligne']
df.head()
```

```python
# Vérification des types des colonnes
df.dtypes
```

```python
# Changement de types
df = df.astype({
    'nom_ligne': 'string',
    'famille_ligne': 'string'
})
df.dtypes
```

```python
# Nombre de lignes au total
df['id_ligne'].nunique()
```

```python
# Classer les lignes par ordre croissant d'id
df = df.sort_values(by = 'id_ligne')
df.head(30)
```

```python
# Export as a csv table
df.to_csv('stg_star_bus_lignes.csv', sep=';', index=False)
```
